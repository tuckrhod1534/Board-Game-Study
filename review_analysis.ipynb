```python
import pickle as pk
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

**Analyze positive sentiments first**

```python
with open('pos_sent.pkl', 'rb') as file:
    pos_sent = pk.load(file)
```

```python
pos_keyword = []

for index, row in pos_sent.iterrows():
    pos_keyword.extend(row.KEY_PHRASES)

unduplicated_phrases = set(pos_keyword)

plt.hist(unduplicated_phrases)
plt.show()
```
